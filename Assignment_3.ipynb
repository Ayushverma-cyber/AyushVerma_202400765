```python
sales_data = [
    ("Laptop", "Electronics", 5),
    ("Mobile", "Electronics", 10),
    ("Headphones", "Electronics", 8),
    ("Shirt", "Clothing", 15),
    ("Jeans", "Clothing", 10),
    ("Shoes", "Clothing", 12),
    ("Table", "Furniture", 6),
    ("Chair", "Furniture", 10),
    ("Sofa", "Furniture", 4),
    ("Laptop", "Electronics", 3),
    ("Shirt", "Clothing", 5),
    ("Chair", "Furniture", 7)
]

columns = ["Product", "Category", "Quantity"]

print("Sales data created successfully!")

for row in sales_data:
    print(row)
```

```python
from pyspark.sql import SparkSession

# Create Spark Session
spark = SparkSession.builder \
    .appName("SalesCategoryAnalysis") \
    .getOrCreate()

print("Spark Session created successfully!")
print("Spark Version:", spark.version)
```

```python
# Create Spark DataFrame
sales_df = spark.createDataFrame(sales_data, columns)

print("Spark DataFrame created successfully!")

# Display the sales data
sales_df.show()
```

```python
from pyspark.sql.functions import sum

# Calculate total products sold in each category
category_sales = sales_df.groupBy("Category") \
    .agg(sum("Quantity").alias("Total_Products_Sold"))

print("Category-wise product sales calculated successfully!")

# Display the result
category_sales.show()
```

```python
# Sort the result by Category
final_result = category_sales.orderBy("Category")

print("Final category-wise sales result:")
final_result.show()
```

```python
# Convert Spark result to Pandas DataFrame
final_result_df = final_result.toPandas()

print("Final Result:")
display(final_result_df)
```

```python
# Calculate grand total of products sold
grand_total = final_result_df["Total_Products_Sold"].sum()

print("Grand Total Products Sold:", grand_total)
```

```python
# Save the final Spark result as a CSV file
final_result.coalesce(1).write.mode("overwrite").option("header", "true").csv("category_sales_result")

print("Final result saved successfully!")
print("Output folder: category_sales_result")
```
