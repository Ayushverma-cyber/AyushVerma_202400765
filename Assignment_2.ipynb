```python
# Assignment 2
# Step 1: Create sample student grade data

data = """StudentID,Subject,Grade
202400001,DBMS,A
202400002,DBMS,B
202400003,DBMS,S
202400004,DBMS,C
202400005,DBMS,B
202400006,DBMS,F
202400007,DBMS,A
202400008,DBMS,D
202400009,DBMS,C
202400010,DBMS,E
"""

with open("student_grades.csv", "w") as file:
    file.write(data)

print("Sample input file created successfully!")
print("\nInput data:")
print(data)
```

```python
# Step 2: Mapper Function

def mapper(record):
    """
    Mapper reads one student record
    and emits (Grade, 1)
    """

    student_id, subject, grade = record.strip().split(",")

    return (grade, 1)


# Test the mapper
with open("student_grades.csv", "r") as file:
    next(file)  # Skip header

    print("Mapper Output:")

    for line in file:
        print(mapper(line))
```

```python
# Step 3: Reducer Function

def reducer(mapped_data):
    """
    Reducer groups grades and calculates
    the total number of students in each grade.
    """

    result = {}

    for grade, count in mapped_data:
        if grade not in result:
            result[grade] = 0

        result[grade] += count

    return result


# Run Mapper
mapped_data = []

with open("student_grades.csv", "r") as file:
    next(file)  # Skip header

    for line in file:
        mapped_data.append(mapper(line))


# Run Reducer
reduced_data = reducer(mapped_data)

print("Reducer Output:")

for grade in sorted(reduced_data):
    print(grade, ":", reduced_data[grade])
```

```python
import pandas as pd

df = pd.read_csv("semester_results_complete.csv")

print("Complete semester dataset loaded successfully!")
print("Total records:", len(df))

print("\nFirst 10 records:")
display(df.head(10))
```

```python
def mapper(record):
    """
    Mapper emits (Subject, Grade)
    for each student record.
    """
    student_id, subject, grade = record
    return (subject, grade)


mapped_data = []

for _, row in df.iterrows():
    mapped_data.append(
        mapper((row["StudentID"], row["Subject"], row["Grade"]))
    )

print("Mapper executed successfully!")

print("\nFirst 15 Mapper outputs:")
for item in mapped_data[:15]:
    print(item)
```

```python
from collections import defaultdict

# Shuffle and Sort phase
grouped_data = defaultdict(list)

for subject, grade in mapped_data:
    grouped_data[subject].append(grade)

print("Shuffle and Grouping completed successfully!")

print("\nGrouped data:")
for subject, grades in grouped_data.items():
    print(subject, ":", grades[:10], "...")
```

```python
from collections import Counter

# Reducer
reduced_data = {}

for subject, grades in grouped_data.items():
    grade_count = Counter(grades)

    reduced_data[subject] = {
        "S": grade_count.get("S", 0),
        "A": grade_count.get("A", 0),
        "B": grade_count.get("B", 0),
        "C": grade_count.get("C", 0),
        "D": grade_count.get("D", 0),
        "E": grade_count.get("E", 0),
        "F": grade_count.get("F", 0)
    }

print("Reducer executed successfully!")

print("\nGrade Count for Each Subject:")
for subject, counts in reduced_data.items():
    print("\n", subject)
    for grade, count in counts.items():
        print(grade, ":", count)
```

```python
# Convert reducer output into a DataFrame
result_df = pd.DataFrame.from_dict(reduced_data, orient="index")

# Give the index a proper name
result_df.index.name = "Subject"

# Display the final result
print("Final MapReduce Result:")
display(result_df)
```
